```python
scores=[90,85,77]
scores[1]=88
scores[2]=80
print(scores)
scores.append(100)#在list末尾添加元素
print(scores)
print(len(scores))#计算list的长度，列表里有几个元素
print(max(scores))
print(min(scores))
print(sum(scores))
average_scores=sum(scores)/len(scores)
print(average_scores)
```

```python
students=[]
print("====学生成绩管理器====")
student_count=int(input("请输入学生人数："))
print(f"请输入学生人数：{student_count}")
for i in range(1,student_count+1):#不能直接遍历一个整数，必须加上range
    name=input("请输入学生姓名：")
    math_score=float(input("请输入数学成绩："))
    english_score=float(input("请输入英语成绩："))
    Python_score=float(input("请输入Python成绩："))
    if math_score<0 or english_score<0 or Python_score<0 or math_score>100 or english_score>100 or Python_score>100:
        print("输入失败！成绩必须在0~100！")
    else:
        print(f"====学生{i}====")
        print(f"姓名：{name}")
        print(f"数学：{math_score}")
        print(f"英语：{english_score}")
        print(f"Python：{Python_score}")
        student={
           "name":name,
           "math":math_score,
           "english":english_score,
           "python":Python_score
        }
        students.append(student)
print("====成绩报告====")
for student in students:
             average_scores=(student["math"]+student["english"]+student["python"])/3
             max_score=max(student["math"],student["english"],student["python"])
             min_score=min(student["math"],student["english"],student["python"])
             print(f"{student['name']}")
             print(f"平均分：{average_scores:.2f}")
             print(f"最高分：{max_score}")
             print(f"最低分：{min_score}")
print("====查找学生====")
student_seek=input("请输入要查找的学生姓名：")
found=False
for student in students:
    if student_seek==student["name"]:
        print(student)
        found=True
if found==False:
    print("查询失败！没有找到该学生！")
```
