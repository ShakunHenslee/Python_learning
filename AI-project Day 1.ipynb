```python
print("====学生成绩分析器====")
name=input("请输入姓名：")
math_score=float(input("请输入数学成绩："))
english_score=float(input("请输入英语成绩："))
python_score=float(input("请输入Python成绩："))
if math_score>100 or math_score<0 or english_score>100 or english_score<0 or python_score>100 or python_score<0:
    print("输入失败，成绩必须在0~100之间！")
else:
    print("成绩输入成功！")
    print("====成绩报告====")
    print(f"姓名：{name}")
    print(f"数学：{math_score}")
    print(f"英语：{english_score}")
    print(f"python:{python_score}")
    p_score=(math_score+english_score+python_score)/3
    max_score=max(math_score,english_score,python_score)
    min_score=min(math_score,english_score,python_score)
    print(f"平均分：{p_score:.2f}")#保留两位小数
    print(f"最高分：{max_score}")
    print(f"最低分：{min_score}")
    if(p_score>=90):
       print("综合评价：优秀")
    elif(p_score>=80):
       print("综合评价：良好")
    elif(p_score>=60):
       print("综合评价：及格")
    else:
       print("综合评价：不及格")

```
