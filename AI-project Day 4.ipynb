```python
import json
def load_students():
     try:
          with open("students.json", "r", encoding="utf-8") as file:
            students=json.load(file)
            print(students)
            return students
     except FileNotFoundError:
         return []
     except json.JSONDecodeError:
        print("数据文件格式错误！")
        return []
def save_students(students):
    with open("students.json", "w", encoding="utf-8") as file:
        json.dump(students, file, ensure_ascii=False, indent=4)
def input_student(i):
    name=input("请输入姓名：")
    try:
       math_score=float(input("请输入数学成绩："))
       english_score=float(input("请输入英语成绩："))
       python_score=float(input("请输入python成绩："))
    except ValueError:
        print("输入失败！成绩必须是数字！")
        return None
    if math_score>100 or english_score>100 or python_score>100 or math_score<0 or english_score<0 or python_score<0:
        print("输入失败！")
        return None
    else:
        print(f"====学生{i+1}====")
        print(f"姓名：{name}")
        print(f"数学：{math_score}")
        print(f"英语：{english_score}")
        print(f"Python：{python_score}")
        student={
        "name":name,
        "math_score":math_score,
        "english_score":english_score,
        "python_score":python_score
    }
        return student
def calculate_score(student):#表示计算一个学生，传入一个学生计算一个学生，所以不需要写遍历，逻辑重复
        average_score=(student["math_score"]+student["english_score"]+student["python_score"])/3
        max_score=max(student["math_score"],student["english_score"],student["python_score"])
        min_score=min(student["math_score"],student["english_score"],student["python_score"])
        return average_score,max_score,min_score
def show_report(students):
    print("====成绩报告====")
    for student in students:
        average_score,max_score,min_score=calculate_score(student)#调用计算函数，并接收它返回的三个结果
        print(f"{student['name']}")
        print(f"平均分：{average_score:.2f}")
        print(f"最高分：{max_score}")
        print(f"最低分：{min_score}")
def search_student(students):
    print("====查找学生====")
    student_seek=input("请输入要查找的学生姓名：")
    found=False
    for student in students:
        if student_seek==student["name"]:
            print(student)
            found=True
    if found==False:
         print("查询失败！没有找到该学生！")
def main():
   students=load_students()
   print("====学生成绩管理器====")
   student_count=int(input("请输入学生人数："))
   for i in range(student_count):
       student=input_student(i)#调用输入函数，把函数返回的学生保存到 student
       if student!=None:
            students.append(student)
            save_students(students)
   show_report(students)#把整个学生列表交给成绩报告函数处理。
   search_student(students)
main()
```
